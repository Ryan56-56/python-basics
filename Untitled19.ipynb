```python
person = {
    'first_name' : 'John',
    'last_name' : 'Doe',
    'age' : 25,
    'country' : 'USA',
    'is_married' : False,
    'skills' : ['JavaScript', 'React', 'Node', 'MongoDB', 'Python', 'Django'],
    'address' : {
        'street' : '123 Main St',
        'zip_code' : '10001'
    }
    }
print(len(person))
print(person['skills'])
print(person['skills'][0])
print(person['skills'][1])
print(person['skills'][2])
print(person['skills'][3])
print(person['skills'][4])
print(person['skills'][5])
print(person['first_name'], person['last_name'])
print(person['country'], person['address'])
person.get('first_name')
person.get('country')
person.get('skills')
person['job_title'] = 'Instructor'
person['skills'].append('HTML') # add a string
print(person)
person['age'] = 22 # changed age to 22
print(person)
print('last_name' in person) #True
print('salary' in person) #False
person.pop('first_name') # removes first name
person.popitem() # removes address item
del person['is_married'] # removes is married
print(person)
print(person.items())
values = person.values()
print(values)

#del person
```
