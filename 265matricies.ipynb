```python
import numpy as np

# Matrix A
A = np.array([
    [2, 5],
    [1, 3]
])

det_A = np.linalg.det(A)

print("Matrix A:")
print(A)
print("Determinant =", det_A)

if abs(det_A) < 1e-10:
    print("A is singular (not invertible)")
else:
    print("A is nonsingular (invertible)")
```

```python
import numpy as np

# Matrix B
B = np.array([
    [2, 5],
    [2, 5]
])

det_B = np.linalg.det(B)

print("Matrix B:")
print(B)
print("Determinant =", det_B)

if abs(det_B) < 1e-10:
    print("B is singular (not invertible)")
else:
    print("B is nonsingular (invertible)")
```

```python
import numpy as np

# Matrix C
C = np.array([
    [1, 1, 3],
    [-2, 1, 0],
    [1, 0, -1]
])

det_C = np.linalg.det(C)

print("Matrix C:")
print(C)
print("Determinant =", det_C)

if abs(det_C) < 1e-10:
    print("C is singular (not invertible)")
else:
    print("C is nonsingular (invertible)")
```

```python
import numpy as np

A = np.array([
    [1, -1, 5],
    [-2, 0, 0],
    [1, 5, -1]
])

print("Matrix A:")
print(A)

try:
    A_inv = np.linalg.inv(A)
    print("\nInverse of A:")
    print(A_inv)
except np.linalg.LinAlgError as e:
    print("\nError:", e)
```

```python
import numpy as np

B = np.array([
    [2, -1],
    [10, -7],
    [5, 2]
])

print("Matrix B:")
print(B)

try:
  B_inv = np.linalg.inv(B)
  print("\nInverse of B:")
  print(B_inv)
except np.linalg.LinAlgError as e:
  print("\nError:", e)
```
