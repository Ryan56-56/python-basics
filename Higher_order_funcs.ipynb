```python
def sum_numbers(nums): # normal function
  return sum(nums) # a sad function abusing the built-in sum function :<
```

```python
def higher_order_function(f, lst):
  summation =f(lst)
  return summation
result = higher_order_function(sum_numbers, [1, 2, 3, 4, 5])
print(result)
```

```python
def cube(x): # a cube function
  return x ** 3
```

```python
def square(x):  # a square function
  return x ** 2
```

```python
def absolute(x):        # an absolute value function
  if x >= 0:
    return x
  else:
    return -(x)
def higher_order_function(type): # a higher order function returning a function
  if type == 'square':
    return square
  elif type == 'cube':
    return cube
  elif type == 'absolute':
    return absolute

result = higher_order_function('square')
print(result(3))
result = higher_order_function('cube')
print(result(3))
result = higher_order_function('absolute')
print(result(-3))
```

```python
def add_ten():
  ten = 10
  def add(num):
    return ten + num
  return add

closure_result = add_ten()
print(closure_result(5))
print(closure_result(10))
```

```python
'''This decorator function is a higher order function
that takes a function as a parameter'''

def uppercase_decorator(function):
  def wrapper():
    func = function()
    make_uppercase = func.upper()
    return make_uppercase
  return wrapper
@uppercase_decorator
def greetings():
  return 'Hello my name is'
print(greetings())
```

```python
'''These decorator functions are higher order functions
that take functions as parameters'''

# First Decorator
def uppercase_decorator(function):
    def wrapper():
        func = function()
        make_uppercase = func.upper()
        return make_uppercase
    return wrapper

# Second decorator
def split_string_decorator(function):
    def wrapper():
        func = function()
        splitted_string = func.split()
        return splitted_string
    return wrapper
#Decorators will be executed from bottom to top
@split_string_decorator
@uppercase_decorator     # order with decorators is important in this case - .upper() function does not work with lists
def greeting():
    return 'Welcome to Python'
print(greeting())   # ['WELCOME', 'TO', 'PYTHON']
```

```python
def decorator_with_parameters(function):
  def wrapper_accepting_parameter(para1, para2, para3):
    function(para1, para2, para3)
    print('I live in {}'.format(para3))
  return wrapper_accepting_parameter

@decorator_with_parameters
def print_full_name(first_name, last_name, country):
  print('I am {} {}. I love anime.'.format(first_name, last_name))

print_full_name('John', "Doe", 'Mexico')
```

```python
# syntax
#map(function, iterable)
numbers = [1, 2, 3, 4, 5] # iterable
def square(x):
  return x ** 2
numbers_squared = map(square, numbers)
print(list(numbers_squared))

```

```python
numbers_squared = map(lambda x : x ** 2, numbers)
print(list(numbers_squared))
```

```python
number_string = ['1', '2', '3', '4', '5']
numbers_int = map(int, number_string)
print(list(numbers_int))
```

```python
names = ['apolo', 'infernious', 'abramas', 'drifter']
def change_to_upeper(name):
  return name.upper()

names_upper_cased = map(change_to_upeper, names)
print(list(names_upper_cased))

names_upper_cased = map(lambda name: name.upper(), names)
print(list(names_upper_cased))
```

```python
numbers = [1, 2, 3, 4, 5]  # iterable

def is_even(num):
  if num % 2 == 0:
    return True
  return False
even_numbers = filter(is_even, numbers)
print(list(even_numbers))
```

```python
numbers = [1, 2, 3, 4, 5]

def is_odd(num):
  if num % 2 != 0:
    return True
  return False
odd_numbers = filter(is_odd, numbers)
print(list(odd_numbers))
```

```python
names = ['apolo', 'infernious', 'abramas', 'drifter']
def is_name_long(name):
  if len(name) > 7:
    return True
  return False
long_names = filter(is_name_long, names)
print(list(long_names))
```

```python
from functools import reduce

number_str = ['1', '2', '3', '4', '5']
def add_two_numbers(x, y):
  return int(x) + int(y)

total = reduce(add_two_numbers, number_str)
print(total)
```
