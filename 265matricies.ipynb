```python
import numpy as np

# Matrix A
A = np.array([
    [2, 5],
    [1, 3]
])

det_A = np.linalg.det(A)

print("Matrix A:")
print(A)
print("Determinant =", det_A)

if abs(det_A) < 1e-10:
    print("A is singular (not invertible)")
else:
    print("A is nonsingular (invertible)")
```

```python
import numpy as np

# Matrix C
C = np.array([
    [1, 1, 3],
    [-2, 1, 0],
    [1, 0, -1]
])

det_C = np.linalg.det(C)

print("Matrix C:")
print(C)
print("Determinant =", det_C)

if abs(det_C) < 1e-10:
    print("C is singular (not invertible)")
else:
    print("C is nonsingular (invertible)")
```
