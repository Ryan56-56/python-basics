```python
assignments = int(input('Enter number of assignments: '))
total = 0 #added
for i in range(assignments): # for the ability to enter each assignment given
  print(f'\n--- Assignment {i + 1} ---')
  grade = float(input('Enter your grade: '))
  total = total + grade
  if grade >= 90:
    print('A')
  elif grade >= 80:
    print('B')
  elif grade >= 70:
    print('C')
  elif grade >= 60:
    print('D')
  else:
    print('F')
average = total / assignments
print(average)
if average >= 90:
    course_grade = 'A'
elif average >= 80:
    course_grade = 'B'
elif average >= 70:
    course_grade = 'C'
elif average >= 60:
    course_grade = 'D'
else:
    course_grade = 'F'
print('Your grade for this course is:', course_grade)
```

```python
user = input('what month is it? ') # in checks wheather a value exists in a group of values
if user in ('September', 'October', 'November'):
  print("The season is autumn")
elif user in ('December', 'January', 'February'):
  print("The season is winter")
elif user in ('March', 'April', 'May'):
  print("The season is spring")
elif user in ('June', 'July', 'August'):
  print("The season is summer")
```

```python
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.append('lime') # adds lime
print("Current fruits:", fruits) # prints current list

fruit = input('Enter a fruit: ') # prompts user to enter a fruit for the checking process

if fruit in fruits: # if the fruit already exist
  print('This fruit already exists in the list')
else: # if it dosent
  fruits.append(fruit) # add the fruit that dont exist
  print('This fruit does not exist in the list. Added to the list!')
  print("Updated fruits:", fruits) # updated print
```
