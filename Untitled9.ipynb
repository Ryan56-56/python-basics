```python
#simple function
def add(a: float, b: float) -> float:
  print(f'adding: {a} + {b}')
  return a + b

print(add(a=10, b=15))
print(add(a=15, b=30))

def greet(name: str, greeting: str ='hi') -> None:
  print(f'{greeting} {name}')

greet(name='Ryan', greeting='Hello')
greet(name='Ryan')
greet(greeting='Hello', name='Ryan')

def get() -> int:
  # Initialize counter attribute if it doesn't exist
  if not hasattr(get, 'calls'):
    get.calls = 0
  get.calls += 1
  print(f'get() has been called {get.calls} times!')
  return get.calls

# Let's test calling it a few times
get()
get()
get()
```
