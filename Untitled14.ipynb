```python
anime = ['one_peice', 'naruto', 'bleach', 'dragon_ball', 'hunter_x_hunter']
anime.sort() # sorts the list in alphabet order
print(anime)
anime.sort(reverse=True) # sorts the list in reverse alphabet order
print(anime)

print(sorted(anime)) # sorts the list in alphabet order
print(sorted(anime))

anime = sorted(anime,reverse=True) # sorts the list in reverse alphabet order
print(anime)
```
