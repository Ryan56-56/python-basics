```python
from datetime import datetime
now = datetime.now()
print(now)
day = now.day
month = now.month
year = now.year
hour = now.hour
minute = now.minute
second = now.second
timstamp = now.timestamp()
print(day , month, year, hour, minute)
print('TimeStamp:', timstamp)
print(f'{day}/{month}/{year}, {hour}:{minute}')
```

```python
from datetime import datetime
new_year = datetime(2020, 1, 1)
print(new_year)
day = new_year.day
month = new_year.month
year = new_year.year
hour = new_year.hour
minute = new_year.minute
second = new_year.second
print(day, month, year, hour, minute)
print(f'{day}/{month}/{year}, {hour}:{second}')
```

```python
from datetime import datetime
# current date time
now = datetime.now()
t = now.strftime("%H:%M:%Y")
print("time:", t)
time_one = now.strftime("%m:%d:%Y, %H:%M:%Y")
print("time one:", time_one)

# lower case y prints 26 instead of 2026
# dd/mm/YY H:M:S format
time_two = now.strftime("%d:%m:%Y, %H:%M:%S")
print("time_two:", time_two)
```

```python
from datetime import datetime
date_string = '5 December, 2019'
print("date_sting", date_string)
date_object = datetime.strptime(date_string, "%d %B, %Y")
print("date_object =", date_object)
```

```python
from datetime import date
d = date(2020, 1, 1)
print(d)
print("current date:", d.today())

today = date.today()
print("Current year:", d.year)
print("Current month:", d.month)
print("Current day:", d.day)

print("Current year:", today.year)
print("Current month:", today.month)
print("Current day:", today.day)
```

```python
from datetime import time
a = time()
print('a = ',a )
b = time(10, 20, 30)
print("b = ",b)
c = time(hour=10, minute=20, second=30)
print('C = ',c)
d = time(10, 20, 50, 20004)
print('d = ',d)
```

```python
from datetime import date, datetime
today = date(year=2026, month=11, day=3)
new_year =date(year=2027,month=1,day=1)
time_left_for_new_year = new_year - today
print('Time left for new year: ',time_left_for_new_year)

t1 = datetime(year=2026, month=11, day=3, hour = 0, minute = 59, second = 0)
t2 =datetime(year=2027,month=1,day=1,hour = 0, minute = 0, second = 0)
diff = t2 - t1
print('Time left for new year: ',diff)
```

```python
from datetime import timedelta
t1 = timedelta(weeks=10, days=5,hours=13,seconds=30)
t2 = timedelta(days=7,hours=13,minutes=30,seconds=30)
t3 = t1 - t2
print("t3 = ",t3)
```
