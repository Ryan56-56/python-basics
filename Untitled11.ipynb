```python
radius = 20
area_of_circle = 3.14 * radius ** 2
print('Area of a circle:', area_of_circle)

#rectangle
length = 10
width = 20
area_of_rectangle = length * width
print("Area of a Rectangle:", area_of_rectangle)

#weight of an object
Mass = 75
Gravity = 9.81
weight = Mass * Gravity
print(weight, 'N')

#density
mass = 75  #in kg
volume = 0.075 # in cubic meters
desity = mass / volume
print(desity, 'kg/m^3')


# == Equal
# != Not equal
# <, <= less
# >, >= more

print(2 != 3)
print(len('mango') == len('avocado'))

print('B not in Asabeneh', 'B' in 'Asabeneh') # False - there is no uppercase B
print('coding' in 'coding for all') # True - because coding for all has the word coding
print(3 > 2 and 4 > 5)
print(3 > 2 or 4 > 5)
print(not 2 == 3)
print(not True)
print(not False)
print(9.8 == 10)
```
